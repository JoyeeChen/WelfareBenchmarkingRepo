```python
from inspect_evals.assistant_bench import assistant_bench_web_browser

from inspect_ai import eval
from inspect_ai.agent import Agent, agent, react
from inspect_ai.tool import web_search

@agent
def web_surfer_2() -> Agent:
    return react(
        name="web_surfer",
        description="Web research assistant",
        prompt="You are an expert at using web search to answer questions.",
        tools=[web_search()],
    )

eval(
    assistant_bench_web_browser(),
    solver=web_surfer_2(),
    model="openai/gpt-4o",
)
```

This dismal score of 0.006 (albeit at a remarkably low cost of around 30 cents as WandB says) begs the question, does adding more tools to the user's repertoire increase the score? Let's add the computer() tool at https://inspect.aisi.org.uk/tools-standard.html#sec-computer but note that computer() requires a model capable of image input! Which openai/gpt-4o thankfully has.

```python
from inspect_evals.assistant_bench import assistant_bench_web_browser

from inspect_ai import eval
from inspect_ai.agent import Agent, agent, react
from inspect_ai.tool import web_search, computer

@agent
def web_surfer_3() -> Agent:
    return react(
        name="web_surfer",
        description="Web research assistant",
        prompt="You are an expert at using web search to answer questions.",
        tools=[web_search(), computer()],
    )

eval(
    assistant_bench_web_browser(),
    solver=web_surfer_3(),
    model="openai/gpt-4o",
)
```

Well at least the accuracy rose to 0.023 I guess? But one natural deficiency I was thinking about was that I just added computer() to the tools section of the react agent, without actually changing the other prompts or other fields in that agent's definition! 

```python
from inspect_evals.assistant_bench import assistant_bench_web_browser

from inspect_ai import eval
from inspect_ai.agent import Agent, agent, react
from inspect_ai.tool import web_search, computer

@agent
def web_surfer_4() -> Agent:
    return react(
        name="web_surfer",
        description="Web research assistant with computer-use capabilities",
        prompt="You are an expert at using web search to answer questions. When you can't find an answer by web search alone, you are able to use a computer desktop environment along with the ability to view the screen and perform mouse and keyboard gestures.",
        tools=[web_search(), computer()],
    )

eval(
    assistant_bench_web_browser(),
    solver=web_surfer_4(),
    model="openai/gpt-4o",
)
```
